# 🏆 4. Sorting and Ranking

Sort and Rank data in a DataFrame by using:
- `.sort_values()` — Sort by one or more column values
- `.sort_index()` — Sort by the row index
- `.rank()` — Assign ranks to numeric values

```python
df_grade = pd.DataFrame(data)

df_grade.loc[3, "Percentage"] = 78 # Duplicate percentage to test tied ranks
df_grade.loc[4, "Percentage"] = 78 # Duplicate percentage to test tied ranks
df_grade.loc[6, "Percentage"] = 78 # Duplicate percentage to test tied ranks
```

### 🏅 Add Ranking Based on Percentage
The `.rank()` function assigns a rank to each value.  
Ties result in averaged ranks by default.

Since the Ranking 1-4 has the same Percentage value >> All of them have the same Ranking :: (1+2+3+4)/4 = 2.5

```python
df_grade['Ranking'] = df_grade['Percentage'].rank(ascending=1)
df_grade
```

`rank(method='dense')` 
- The `dense` method assigns the same rank to ties, but keeps the next rank sequential.

```python
df_grade['Rank_Dense'] = df_grade['Percentage'].rank(method='dense')
df_grade
```

### 🔢 Setting a different column as index can help in sorting or indexing.

inplace: Boolean; if True, modifies the original DataFrame

`df_grade = df_grade.set_index('Ranking')` == `df_grade.set_index('Ranking', inplace=True)`

```python
df_grade = pd.DataFrame(data)
df_grade.loc[3, "Percentage"] = 78 # Duplicate percentage to test tied ranks
df_grade['Ranking'] = df_grade['Percentage'].rank(ascending=1)

df_grade.set_index('Ranking', inplace=True)
print(df_grade)
```

Sort by Index (Row Labels)

```python
df_grade.sort_index()
```

### 🔃 Sort by Column

```python
df_grade.sort_values(by='Age', ascending= False)
```

Sort by Multiple Columns
- Sort by `Age` descending, and break ties with `Ranking` ascending.

```python
df_grade.sort_values(by=['Age', 'Ranking'], ascending= [False, True], inplace=True)
print(df_grade)
```

- Use `na_position='first'` or `'last'` to control where `NaNs` appear

By default, `sort_values()` places `NaN` values at the end of the sorted result.  
You can control this behavior using the `na_position` parameter.

- `na_position='first'` puts missing values at the top
- `na_position='last'` (default) puts them at the bottom

```python
df_grade.iloc[3,1] = np.nan
df_grade
```

```python
print("Sort by Percentage (NaNs last - default):")
print(df_grade.sort_values(by='Age'))

print("\nSort by Percentage (NaNs first):")
print(df_grade.sort_values(by='Age', na_position='first'))
```

# 🧪 5. Conditional Column Creation & Mapping

This section covers how to:
- Use `.map()` with custom functions or dictionaries
- Use `np.where()` to conditionally assign values
- Use `.loc[]` for condition-based updates
- Fill missing values with `.fillna()`

```python
df_grade = pd.DataFrame(data)
df_grade
```

### Use a Custom Function with `.map()` to Categorize Grades

```python
def grade_stats(value):
    if value < 50:
        return "very low"
    if 50 <= value < 65:
        return "low"
    elif 65 <= value < 75:
        return "average"
    elif 75 <= value < 90:
        return "better"
    elif value >= 90:
        return "very good"

df_grade['grade_stats'] = df_grade['Percentage'].map(grade_stats)
df_grade
```

### Map Strings to New Values Using a Dictionary

```python
mappings = {
    'Math': 'Building_1',
    'Science': 'Building_2'
}

df_grade['Building'] = df_grade['Stream'].map(mappings)
df_grade
```

- Handle Missing Values with `.fillna()`
- Replace NaNs (from unmapped categories) with `'Others'`.

```python
df_grade = df_grade.fillna('Others')
df_grade
```

### Use `np.where()` for Inline Binary Classification
Create a new column based on a condition

```python
df_grade['Age>20'] = np.where(df_grade['Age'] >= 20, 'Yes', 'No')
df_grade
```

### Update Values Conditionally Using `.loc[]`
Change the building label for Science stream students.

```python
df_grade.loc[df_grade['Stream'] == 'Science', 'Building'] = 'Building_3'
df_grade
```

### Aggregating statistics grouped by category

Groups your data by one or more categorical columns, and applies summary stats.

Use when you want to know:
- Average score by stream
- Count per category
- Stats per subgroup (e.g., year & stream)

This function fits in the more general split-apply-combine pattern:
- Split the data into groups
- Apply a function to each group independently
- Combine the results into a data structure

The apply and combine steps are typically done together in pandas.

```python
df_stat[['Stream', 'Percentage']].groupby('Stream').mean()
```

```python
df_stat.groupby('Stream')['Percentage'].mean()
```

🔍 1. `df_stat[['Stream', 'Percentage']].groupby('Stream').mean()`

✅ What it does:
- First `selects only the Stream and Percentage columns.`
- Then groups by Stream.
- Finally, computes the mean for only the numeric columns (in this case just Percentage).

🧠 Why Use This:
- Useful if you want to restrict aggregation only to selected columns before grouping.

📦 Output:
- A `DataFrame` with Stream as the index and Percentage as the column.


🔍 2. `df_stat.groupby('Stream')['Percentage'].mean()`

✅ What it does:
- First `groups the entire DataFrame by Stream.`
- Then selects the Percentage column after grouping.
- Then computes the mean.

🧠 Why Use This:
- More explicit when you're working with a single column.
- Cleaner if you're not planning to include any other column in the aggregation.

📦 Output:
A `Series` — not a DataFrame — where:
- Index is `Stream`
- Values are the mean of `Percentage`

```python
print('Output as DataFrame : ', df_stat.groupby(['Stream','Year'])['Percentage'].mean().iloc[0])
print('Output as Series : ', df_stat[['Stream','Year', 'Percentage']].groupby(['Stream','Year']).mean().iloc[0]['Percentage'])
```

```python
# Find the mean to all column that is numberic only
df_stat.groupby('Stream').mean(numeric_only=True)
```

### 🔗 7. Joining and Merging

This section explores how to combine multiple DataFrames using:
- `pd.concat()` for stacking
- `pd.merge()` for SQL-style joins
- `.join()` for index-based merging

These functions help when you need to merge datasets from different sources or shapes.

### `pd.concat()` — Combine Along Rows or Columns

Similar to append from list
- Append in vertical (add more role) >> axis = 0 : default
- Append in horizontal (add more column) >> axis = 1 

```python
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Alice', 'Bob', 'Charlie']
})

df2 = pd.DataFrame({
    'ID': [4, 5],
    'Name': ['David', 'Eve']
})

# Stack rows (vertical concatenation)
df_concat_rows = pd.concat([df1, df2], ignore_index=True)
print("Concatenate Rows:\n", df_concat_rows)
```

```python
# Stack columns (horizontal concatenation) by using axis=1
df3 = pd.DataFrame({
    'Score': [90, 85, 88]
})

df_concat_cols = pd.concat([df1, df3], axis=1)
print("Concatenate Columns:\n", df_concat_cols)
```


| Feature                   | `.join()`                                          | `pd.merge()`                                |
| ------------------------- | -------------------------------------------------- | ------------------------------------------- |
| **Primary use**           | Index-based joins                                  | SQL-style joins on column(s)                |
| **Default key**           | Joins on the **index** of the calling DataFrame    | Joins on a **common column** (`on='col'`)   |
| **Ease of use**           | Simpler when indexes are already aligned           | More flexible and explicit                  |
| **Join types**            | Supports `'left'`, `'right'`, `'outer'`, `'inner'` | Same (`how=` parameter)                     |
| **Syntax style**          | Method: `df1.join(df2)`                            | Function: `pd.merge(df1, df2, on='col')`    |
| **Multiple keys**         | ❌ Not directly supported                         | ✅ Yes (`on=['key1', 'key2']`)              |
| **Column name conflicts** | Needs handling with `lsuffix`, `rsuffix`           | Can handle duplicate columns more elegantly |


### `pd.merge()` — SQL-style Joins
- Combine on column keys (SQL)
- Based on Column Values

```python
students = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Class': ['A', 'B', 'A', 'C']
})

scores = pd.DataFrame({
    'ID': [1, 2, 4, 5],
    'Score': [90, 85, 88, 70]
})
```

```python
# Inner Merge (default)
print("Inner Merge:\n", pd.merge(students, scores, on='ID'))
```

```python
# Left Merge
print("Left Merge:\n", pd.merge(students, scores, on='ID', how='left'))
```

```python
# Right Merge
print("Right Merge:\n", pd.merge(students, scores, on='ID', how='right'))
```

```python
# Outer Merge
print("Outer Merge:\n", pd.merge(students, scores, on='ID', how='outer'))
```

```python
students = pd.DataFrame({
    'ID': [1, 2, 3, 4, 5, 6],
    'Name': ['A', 'B', 'C', 'D', 'E', 'F'],
    'Class': ['Math', 'Sci', 'Art', 'Sci', 'Math', 'Sci']
})

sci_scores = pd.DataFrame({
    'ID': [1, 2, 4, 5, 7, 8],
    'Score': [90, 85, 88, 70, 60, 52],
    'Class': ['Sci', 'Sci', 'Sci', 'Sci', 'Sci', 'Sci']
})

# Merge with Multiple keys
print("Inner Merge:\n", pd.merge(students, sci_scores, on=['ID', 'Class']))
```

### `.join()` — Simpler Index-Based Join
- Combine on index
- Based on Index

```python
grades = pd.DataFrame({
    'Grade': ['A', 'B', 'C', 'D'],
}, index=['ID1', 'ID2', 'ID3', 'ID4'])

comments = pd.DataFrame({
    'Comment': ['Excellent', 'Good', 'Average', 'Bad']
}, index=['ID1', 'ID2', 'ID4', 'ID5'])

# Join on index (how >> Default 'left')
print("Join on index:\n",  grades.join(comments))
```

```python
# Right Join
print("Right Join:\n", grades.join(comments, how='right'))
```

```python
# Inner Join
print("Inner Join:\n", grades.join(comments, how='inner'))
```

### 🔁 8. Iteration and Row-wise Operations
.iterrows(), .itertuples()

### `.items()` – Iterate Over Columns
Iterates over each column, returning the column name and the entire column as a Series.

```python
data = {
  "firstname": ["Sally", "Mary", "John"],
  "age": [50, 40, 30]
}

df = pd.DataFrame(data)

for key, value in df.items():
  print('First item (key) : \n',key)
  print('\nSecond item (value) : \n',value)
  print('--------------------------')
```

### `.iterrows()`

The iterrows() method generates an iterator object of the DataFrame, allowing us to iterate each row in the DataFrame.

Each iteration produces an index object and a row object (a Pandas Series object).

>> Iterate Over Rows (Slow, but Common)
>> - Returns each row as an (index, Series) pair.
>> - Use when you need access to row data by column name.

✅ Use .iterrows() when:
- You need access to column names
- You're building a new list or conditionally modifying data


```python
data = {'Item': ['Apple', 'Reduce_A', 'Reduce_B', 'Reduce_c','Banana', 'Orange'],
        'Quantity': [10, 4, 1, 2, 20, 30],
        'Price': [5, -2, -10, -5, 3, 7,]}
df = pd.DataFrame(data)
total_sales = []
# Iterating over rows
for index, row in df.iterrows():
    if row['Price'] > 0:
        total_sales.append(row['Quantity'] * row['Price'])
    else:
        total_sales.append(0)

df['total_sales'] = total_sales
print(df)
```

### `.itertuples()`

The itertuples() method generates an iterator object of the DataFrame, ***returning each row as a Pyton Tuple object***.

>> Faster Row Iteration
>> - Returns each row as a namedtuple — access is done by position (faster than .iterrows()).

✅ Use .itertuples() when:
- You don’t need to modify the DataFrame
- Speed is more important than column name access

```python
data = {
  "firstname": ["Sally", "Mary", "John"],
  "age": [50, 40, 30]
}

df = pd.DataFrame(data)

for row in df.itertuples():
  print(row)
  print('Age : ', row[2])
```

```python
data = {'Item': ['Apple', 'Reduce_A', 'Reduce_B', 'Reduce_c','Banana', 'Orange'],
        'Quantity': [10, 4, 1, 2, 20, 30],
        'Price': [5, -2, -10, -5, 3, 7,]}
df = pd.DataFrame(data)
total_sales = []
# Iterating over rows
for index, row in df.iterrows():
    if row['Price'] > 0:
        total_sales.append(row['Quantity'] * row['Price'])
    else:
        total_sales.append(0)

df['total_sales'] = total_sales
print(df)
```

Selecting rows in pandas DataFrame based on conditions

```python
data = {
  'Name': ['Ankit', 'Amit', 'Aishwarya', 'Priyanka', 'Priya', 'Shaurya' ],
  'Age': [21, 19, 20, 18, 17, 21],
  'Stream': ['Math', 'Commerce', 'Science', 'Math', 'Math', 'Science'],
  'Percentage': [88, 92, 95, 70, 65, 78]
}
df = pd.DataFrame(data)

# selecting rows based on condition
grade_a = df[df['Percentage'] >= 80]
print(grade_a)
```

```python
pos_price = df['Name'][df['Percentage'] >= 80]
print(pos_price)
print('\n')
print(pos_price.tolist())
```

```python
# isin() like 'in' list
options = ['Math', 'Commerce']

# selecting rows based on condition in options
rslt_df = df[df['Stream'].isin(options)]
print('Result dataframe :\n', rslt_df)

# selecting rows based on condition outside options
rslt_df = df[~df['Stream'].isin(options)]

print('\nResult dataframe :\n', rslt_df)
```

```python
# Selecting rows based on multiple column conditions using '&' operator.
rslt_df = df[(df['Age'] == 21) & (df['Stream'].isin(options))]
print(rslt_df)
```

```python
print(df[df['Name'].str.contains('Pri')])
```

# Insert row

```python
# Let's create the dataframe
print(df)
print('\n')
rows, columns = df.shape

# Function to insert row in the dataframe
def Insert_row_(row_number, df, row_value):
    # Slice the upper half of the dataframe
    df1 = df.iloc[0:row_number]
 
    # Store the result of lower half of the dataframe
    df2 = df.iloc[row_number:]
 
    # Insert the row in the upper half dataframe
    df1.loc[row_number]=row_value
 
    # Concat the two dataframes
    df_result = pd.concat([df1, df2])
 
    # Reassign the index labels
    df_result.index = [*range(df_result.shape[0])]
 
    # Return the updated dataframe
    return df_result
 
# Let's create a row which we want to insert
row_number = 2
row_value = ['Add_1', 25, 'Science', 85]

if row_number > rows:
    print("Invalid row_number")
else:

    # Let's call the function and insert the row
    # at the second position
    df = Insert_row_(row_number, df, row_value)

    # Print the updated dataframe
    print(df)
```

# Mapping

```python
def grade_stats(value):
    if value < 50:
        return "very low"
    if 50 <= value < 65:
        return "low"
    elif 65 <= value < 75:
        return "average"
    elif 75 <= value < 90:
        return "better"
    elif value >= 90:
        return "very good"

df['grade_stats'] = df['Percentage_col'].map(grade_stats)
df
```

```python
mappings = {
    'Math': 'Building_1',
    'Science': 'Building_2'
}

df['Building'] = df['Stream_col'].map(mappings)
df
```
